```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'text' : ['people watch campusx', 'campusx watch campusx', 'people write comment', 'campusx write comment'], 'output' : [1,1,0,0]})
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
```

```python
bow = cv.fit_transform(df['text'])
```

```python
# printing the vocabulary (unique words present in the corpus)

print(cv.vocabulary_)
```

```python
print(bow[0].toarray())    # 1st sentence
print(bow[1].toarray())    # 2nd sentence

# so we can see our output as a vector below
```

```python
# take a new sentence and predict for it :-

cv.transform(["campusx watch and write comment of campusx"]).toarray()
```
